## Graphs
- BFS:
    - suitable for shortest path/distance between 2 vertices in unweighted graph. 
    - graph of unknown size, e.g. word ladder, or even infinite size, e.g. knight shortest path
- DFS:
    - suitable exploring nodes far from root, e.g., looking for an exit in a maze.
    - uses less memory than BFS for wide graphs bc BFS has to keep all the nodes in the queue -> q can be quite large.

````python
from collections import deque
class Node:
    def __init__(self, val = 0, neighbors = None):
        self.val = val
        self.neighbors = neighbors if neighbors is not None else []
    def __repr__(self) -> str:
        s = "{ "+str(self.val)+" } -> ["
        for i in self.neighbors:
            s += str(i.val) + ", "
        s+= " ]"
        return s
def toAdjList(adjList: list) -> list:
    nodes = {}
    
    for i, neighbors in enumerate(adjList, start=1):
        nodes[i] = Node(i)

    for i, neighbors in enumerate(adjList, start=1):
        nodes[i].neighbors = [nodes[neighbor] for neighbor in neighbors]

    return nodes

            
````

#### [133. Clone Graph](https://leetcode.com/problems/clone-graph/)

- A dictionary nodes = {oldNodes:newNode}
- DFS recursive():
    - nodes[oldNodes] = dfs(oldNodes)
    - Base: Return nodes[oldNodes] if oldNodes in nodes?
        - Return when see a node already created

````python
from typing import Optional
class Solution:
    def cloneGraph(node: Optional['Node']) -> Optional['Node']:
        cloneNodes = {}
        def dfs(oldNode):
            if oldNode in cloneNodes: #if this oldNode already cloned
                return cloneNodes[oldNode]
            
            cloneNodes[oldNode] = Node(val=oldNode.val)
            
            for i in oldNode.neighbors:
                cloneNodes[oldNode].neighbors.append(dfs(i))
                
            return cloneNodes[oldNode]
            
        return dfs(node)

adjList = toAdjList([[2,4],[1,3],[2,4],[1,3]])
print(adjList)
Solution.cloneGraph(adjList[1])
````

#### [417. Pacific Atlantic Water Flow](https://leetcode.com/problems/pacific-atlantic-water-flow/)

From borders, DFS inward to see what cells go to each ocean. For passing in DFS in the first loop:
- From P up to down, P left to right
- From A down to up, A right left

Then go 4 directions from each cell.

For each cell, if cell is both in A,P, add to ouput

**OG Approach**: O((m.n)^2)
- DFS 4 directions from each cell

````python
#Improvised than Neetcode: starts DFS 1 step inward from borders cuz borders -> auto PA/AT
#Note: Have to add border cell before DFS in the loop in range(CO)/range(RO)
class Solution:
    def pacificAtlantic(self, heights: list[list[int]]) -> list[list[int]]:
        al,pa,out = set(), set(),[]
        RO, CO = len(heights), len(heights[0])
        
        def dfs(r,c, visit ,prevh):

            if (r not in range(RO) or 
                c not in range(CO) or
                (r,c) in visit or
                heights[r][c] < prevh):
                return
            
            visit.add((r,c))
            dfs(r+1, c, visit, heights[r][c])
            dfs(r, c+1, visit, heights[r][c])
            dfs(r-1, c, visit, heights[r][c])
            dfs(r, c-1, visit, heights[r][c])
            return
        
        #All cells could go to Pacific/Atlantic will be added to pa, al sets
        for i in range(CO):
            #Add border to visit + pass border as prevheight (<=)
            pa.add((0,i))
            dfs(1, i, pa , heights[0][i])
            al.add((RO-1,i))
            dfs(RO-2, i, al , heights[RO-1][i])
        for i in range(RO):
            pa.add((i,0))
            dfs(i, 1, pa ,heights[i][0])
            al.add((i, CO-1))
            dfs(i, CO-2, al , heights[i][CO-1])
            
        # print(pa,'\n',al)
        for i in range(RO):
            for j in range(CO):
                if (i,j) in al and (i,j) in pa:
                    out.append([i,j])
        return out    
heights = [[1,2,2,3,5],[3,2,3,4,4],[2,4,5,3,1],[6,7,1,4,5],[5,1,1,2,4]]
Solution.pacificAtlantic(Solution, heights)
````

#### [994. Rotting Oranges](https://leetcode.com/problems/rotting-oranges/description/)
**Approach**: Traverse all neighbors of certain nodes at a time -> BFS

- Go through board to init the BFS queue with all rottens + count freshes
- Start BFS to count minutes and decre freshes.

````python
from collections import deque
class Solution:
    def orangesRotting(self, grid: list[list[int]]) -> int:
        minTime = 0
        q = deque()
        fresh = 0 # need to know #fresh to stop early
        dir = [(0,1),(1,0), (-1,0), (0,-1)]
        #Count fresh and init rotten in q
        for i in range(len(grid)):
            for j in range(len(grid[0])):
                if grid[i][j] == 2:
                    q.append((i,j))
                elif grid[i][j] == 1:
                    fresh += 1
        #question: is visited == turning val 1->2 
        #BFS
        #while q: there's still unprocessed rottens;
        #-> no unprocessed rottens left OR no fresh left(mission done) ->terminate loop
        while q and fresh > 0:
            l = len(q)
            for _ in range(l):
                r,c = q.popleft()
                
                for dr,dc in dir:
                    if (r+dr in range(len(grid)) 
                        and c+dc in range(len(grid[0])) 
                        and grid[r+dr][c+dc] == 1) :
                        
                        grid[r+dr][c+dc] = 2 #same as add to visit
                        q.append((r+dr,c+dc))
                        fresh-=1
            minTime+=1
        return minTime if fresh <=0 else -1
grid = [[2,1,1],[1,1,0],[0,1,1]]           
Solution.orangesRotting(Solution,grid)
````

#### [Valid Tree](https://neetcode.io/problems/valid-tree)
For a undirected TREE cycle detection: 
- All nodes connected + no loop.
- We can traverse a whole undirected tree at ANY NODE, and we CAN visit each node ONLY ONCE (by keeping a prev variable prevent one node to go back)
- Add all nodes to visitSet throughout tree
    - DONT REMOVE visit node after process children like [DIRECTED GRAPH cycle detect](#207-course-schedule)
    - in directed graph, there's case where 2 nodes point to 1 nodeX create 2 dfs path, nodeX must be removed to be revisted in another DFS path
    - in undirected, 3 nodes point to each other, don't have to revisit any node.


````python
class Solution:
    # Use visitSet to detect revisited node (loop)
    # Memorize prevNode when DFS each Node to prevent Node go back, only go forward
    def validTree(self, n: int, edges: list[list[int]]) -> bool:
        graph = defaultdict(list)
        visit = set()
        for n1, n2 in edges:
            graph[n1].append(n2)
            graph[n2].append(n1)
        
        def dfs(node, prev):
            if node in visit:
                return False
            visit.add(node)
            for child in graph[node]:
                if child != prev: #dfs all children except prev (parent)
                    if not dfs(child, node):
                        return False
            return True
        return dfs(0, -1) and len(visit) == n
edges = [[0,1],[0,2],[0,3],[1,4]]
n =5
Solution.validTree(Solution, n, edges)
````

#### [323. Number of Connected Components in an Undirected Graph](https://leetcode.com/problems/number-of-connected-components-in-an-undirected-graph/description/)

##### DFS approach/BFS
- Only use visit to track all nodes, no need prev because prev already in visit which we won't visit again 
    - and Don't care about if cycle exist, so don't separate about prev and other VISITED CHILD like [Valid tree](#valid-tree)
- Overall count++ whenever dfs node not in visit  
    DFS/BFS will traverse to node's all connected children/grandchildren

````python
class Solution:
    def countComponents(self, n: int, edges: list[list[int]]) -> int: 
        graph= defaultdict(list)
        for n1, n2 in edges:
            graph[n1].append(n2)
            graph[n2].append(n1)
        count = 0
        visit = set()

        def dfs(node):
            visit.add(node)
            for child in graph[node]:
                if child not in visit:
                    dfs(child)          
            
        for i in range(n):
            if i not in visit:
                dfs(i) #pass dfs of a new component
                count+=1
        return count
        
n=6
edges=[[0,1], [1,2], [2,3], [4,5]]

Solution.countComponents(Solution, n, edges)
````

#### [1219. Path with Maximum Gold](https://leetcode.com/problems/path-with-maximum-gold)

````python
class Solution:
    def getMaximumGold(self, grid: List[List[int]]) -> int:
        # go 4^k, k - number of gold cells
        #Set cell to 0 to mark as visited temporarily, after exploring all paths from that cell, restores og value -> other paths can reuse the cell.
        # Because this is the longest path finding problem, not just explores all cells problems.
        maxGold = 0 
        m, n = len(grid), len(grid[0])
        directions = [(0, 1), (0, -1), (-1, 0), (1, 0)]
        def dfs(i, j):
            # Base case: 
            if not (0<=i<m and 0<=j<n and grid[i][j] != 0):
                return 0
            
            tempMax = 0
            cellVal = grid[i][j]
            grid[i][j] = 0
            for dr, dc in directions:
                tempMax = max(tempMax, dfs(i + dr, j + dc)) #find the max path among 4 paths
            grid[i][j] = cellVal #restores value to use by other path,
            return cellVal + tempMax #return cell + tempMaxPath to parent
        for i in range(m):
            for j in range(n):
                if grid[i][j] != 0:
                    maxGold = max(maxGold, dfs(i, j))
        return maxGold
````

#### [133. Clone Graph](https://leetcode.com/problems/clone-graph/)

- A dictionary nodes = {oldNodes:newNode}
- DFS recursive():
    - nodes[oldNodes] = dfs(oldNodes)
    - Base: Return nodes[oldNodes] if oldNodes in nodes?

#### [994. Rotting Oranges](https://leetcode.com/problems/rotting-oranges/description/)
**Approach**: Traverse all neighbors of certain nodes at a time -> BFS

- Go through board to init the BFS queue with all rottens + count freshes
- Start BFS to count minutes and decre freshes.

````python
from collections import deque
class Solution:
    def orangesRotting(self, grid: list[list[int]]) -> int:
        minTime = 0
        q = deque()
        fresh = 0 # need to know #fresh to stop early
        dir = [(0,1),(1,0), (-1,0), (0,-1)]
        #Count fresh and init rotten in q
        for i in range(len(grid)):
            for j in range(len(grid[0])):
                if grid[i][j] == 2:
                    q.append((i,j))
                elif grid[i][j] == 1:
                    fresh += 1
        #question: is visited == turning val 1->2 
        #BFS
        #while q: there's still unprocessed rottens;
        #-> no unprocessed rottens left OR no fresh left(mission done) ->terminate loop
        while q and fresh > 0:
            l = len(q)
            for _ in range(l):
                r,c = q.popleft()
                
                for dr,dc in dir:
                    if (r+dr in range(len(grid)) 
                        and c+dc in range(len(grid[0])) 
                        and grid[r+dr][c+dc] == 1) :
                        
                        grid[r+dr][c+dc] = 2 #same as add to visit
                        q.append((r+dr,c+dc))
                        fresh-=1
            minTime+=1
        return minTime if fresh <=0 else -1
grid = [[2,1,1],[1,1,0],[0,1,1]]           
Solution.orangesRotting(Solution,grid)
````

#### [210. Course Schedule II](https://leetcode.com/problems/course-schedule-ii/description/)

````python
class Solution:
    def findOrder(self, numCourses: int, prerequisites: list[list[int]]) -> list[int]:
        topo = []
        degree = [0]*numCourses
        graph = defaultdict(list)
        q = deque()
        #1. Get indegrees and init graph
        for des, src in prerequisites:
            degree[des] +=1
            graph[src].append(des)
            
        #2. Init Q with 0-degree, Q only has 0 degree nodes
        # (node's parent processed in topo)
        for node in range(numCourses):
            if degree[node] == 0:
                q.append(node)
        #3. Go through Q, -1 each visit of non-0 degree child, add to res and Q if child.degree = 0
        while q:
            popNode = q.popleft()
            topo.append(popNode) #Add TOPO, degree always == 0
            for child in graph[popNode]:
                if degree[child] > 0: degree[child] -=1
                if degree[child] == 0: q.append(child)
        if len(topo) == numCourses: return topo
        return []
````

#### [Valid Tree](https://neetcode.io/problems/valid-tree)
For a undirected TREE cycle detection: 
- All nodes connected + no loop.
- We can traverse a whole undirected tree at ANY NODE, and we CAN visit each node ONLY ONCE (by keeping a prev variable prevent one node to go back)
- Add all nodes to visitSet throughout tree
    - DONT REMOVE visit node after process children like [DIRECTED GRAPH cycle detect](#207-course-schedule)
    - in directed graph, there's case where 2 nodes point to 1 nodeX create 2 dfs path, nodeX must be removed to be revisted in another DFS path
    - in undirected, 3 nodes point to each other, don't have to revisit any node.


````python
class Solution:
    # Use visitSet to detect revisited node (loop)
    # Memorize prevNode when DFS each Node to prevent Node go back, only go forward
    def validTree(self, n: int, edges: list[list[int]]) -> bool:
        graph = defaultdict(list)
        visit = set()
        for n1, n2 in edges:
            graph[n1].append(n2)
            graph[n2].append(n1)
        
        def dfs(node, prev):
            if node in visit:
                return False
            visit.add(node)
            for child in graph[node]:
                if child != prev: #dfs all children except prev (parent)
                    if not dfs(child, node):
                        return False
            return True
        return dfs(0, -1) and len(visit) == n
edges = [[0,1],[0,2],[0,3],[1,4]]
n =5
Solution.validTree(Solution, n, edges)
````

    DFS/BFS will traverse to node's all connected children/grandchildren

````python
class Solution:
    def countComponents(self, n: int, edges: list[list[int]]) -> int: 
        graph= defaultdict(list)
        for n1, n2 in edges:
            graph[n1].append(n2)
            graph[n2].append(n1)
        count = 0
        visit = set()

        def dfs(node):
            visit.add(node)
            for child in graph[node]:
                if child not in visit:
                    dfs(child)          
            
        for i in range(n):
            if i not in visit:
                dfs(i) #pass dfs of a new component
                count+=1
        return count
        
n=6
edges=[[0,1], [1,2], [2,3], [4,5]]

Solution.countComponents(Solution, n, edges)
````

#### [1219. Path with Maximum Gold](https://leetcode.com/problems/path-with-maximum-gold)

````python
class Solution:
    def getMaximumGold(self, grid: List[List[int]]) -> int:
        # go 4^k, k - number of gold cells
        #Set cell to 0 to mark as visited temporarily, after exploring all paths from that cell, restores og value -> other paths can reuse the cell.
        # Because this is the longest path finding problem, not just explores all cells problems.
        maxGold = 0 
        m, n = len(grid), len(grid[0])
        directions = [(0, 1), (0, -1), (-1, 0), (1, 0)]
        def dfs(i, j):
            # Base case: 
            if not (0<=i<m and 0<=j<n and grid[i][j] != 0):
                return 0
            
            tempMax = 0
            cellVal = grid[i][j]
            grid[i][j] = 0
            for dr, dc in directions:
                tempMax = max(tempMax, dfs(i + dr, j + dc)) #find the max path among 4 paths
            grid[i][j] = cellVal #restores value to use by other path,
            return cellVal + tempMax #return cell + tempMaxPath to parent
        for i in range(m):
            for j in range(n):
                if grid[i][j] != 0:
                    maxGold = max(maxGold, dfs(i, j))
        return maxGold
````

#### [1319. Number of Operations to Make Network Connected](https://leetcode.com/problems/number-of-operations-to-make-network-connected)

````python
#Regular DFS, BFS
class Solution:
    def makeConnected(self, n: int, connections: List[List[int]]) -> int:
        '''to extract with minimum, found # of disconnected, use dfs or bfs is fine'''
        # Grady cool code bfs and dfs, traverse only to mark visited
        if len(connections) < n - 1: return -1
        visit = set()
        graph = defaultdict(list)
        numOps = 0
        for src, des in connections: # 2way graph
            graph[src].append(des)
            graph[des].append(src)
        def dfs(node):
            if node in visit:
                return
            visit.add(node)
            for child in graph[node]:
                dfs(child)
        def bfs(node):
            q = deque([node])
            while q:
                popNode = q.popleft()
                for child in graph[popNode]:
                    if child not in visit:
                        visit.add(child) # mark visit child avoid processing same node EARLIER than mark when popNode
                        q.append(child)  
        for i in range(n):
            if i not in visit: 
                bfs(i) #or dfs(i) 🔴 NOTICE HERE
                numOps += 1
        return numOps - 1
````

        ans[node] = LeastQNode found, return LeastQNode

Lesson learn: find a temporary min value of each node to their children for every node

````python
class Solution:
    def loudAndRich(self, richer: list[list[int]], quiet: list[int]) -> list[int]:
        graph = defaultdict(list)
        ans = [-1]*len(quiet)
        for des, src in richer:
            graph[src].append(des)
        def dfs(node):
            # if node's child list empty or ans[node] filled
            if graph[node] == [] or ans[node] != -1:
                if ans[node] == -1:
                    ans[node] = node 
                return ans[node]
            leastQNode = node
            #find leastQNode for node among children
            for child in graph[node]:
                childLeastQNode = dfs(child)
                if quiet[childLeastQNode] < quiet[leastQNode]:
                    leastQNode = childLeastQNode
            #after dfs all node's childs
            ans[node] = leastQNode
            return leastQNode
        for node in range(len(quiet)):
            if ans[node] == -1:
                dfs(node)
        return ans
#time + space: O(n^2), n is #pple , graph has n^2 edges  
richer = [[1,0],[2,1],[3,1],[3,7],[4,3],[5,3],[6,3]]
quiet = [3,2,5,4,6,1,7,0]
Solution.loudAndRich(Solution, richer , quiet)
````

````python
from collections import deque  
def floodFill(image: list[list[int]], sr: int, sc: int, color: int) -> list[list[int]]:
    #use visited to prevent inf
    queue = deque([[sr, sc]])
    ROW, COL = len(image), len(image[0])
    visited = set()
    og = image[sr][sc]
    while queue:
        n = len(queue)
        for _ in range(n):
            r, c = queue.popleft()
            image[r][c] = color
            for nrow, ncol in [[r, c+1], [r, c-1], [r-1, c], [r+1, c]]: #neighbor row and col
                if (0<=nrow<ROW and 0<=ncol<COL) and image[nrow][ncol] == og and image[nrow][ncol] != color and (nrow, ncol) not in visited:
                    visited.add((nrow, ncol))
                    queue.append([nrow, ncol])
    return image
                    
image = [[0,0,0],[0,1,0]]
sr = 1
sc = 1
color = 2
floodFill(image, sr, sc, color)
````

````python
from collections import deque
#Time: Space:
#1.  knife is symmetric, and symetric diagonal-> optimize by: abs(x,y), only search first quadrant.
# => only look for neighbor coor in range -2<=neightborR<=x,-2<=neightborC<=y
def get_knight_shortest_path(x: int, y: int) -> int:
    directions = ((2,1),(-2, 1), (2, -1),(-2,-1),(1, 2),(-1, 2),(1, -2), (-1, -2))
    visited = set()
    queue = deque([(0,0)])
    steps = 0
    x,y = abs(x), abs(y)
    while queue:
        n = len(queue)
        for _ in range(n):
            # print(queue)
            r, c = queue.popleft()
            if r == x and c == y:
                return steps
            for dr, dc in directions:
                nrow, ncol = r+dr, c+dc
                if (nrow, ncol) not in visited and -2<=nrow<=x+2 and -2<=ncol<=y+2:
                    visited.add((nrow, ncol))
                    queue.append((nrow, ncol))
        steps+=1
    return steps
                
get_knight_shortest_path(1,1)
````

#### [286. Walls and Gates](https://algo.monster/problems/walls_and_gates)
BFS: O(V + E), n * m vertices and also n * m edges. - >Time: O(n * m).
- Remember: 
    - no need visited set, 
    - when see a visited (non-INF) when we arrived, there must have been a shorter path towards it, so that we can prune the tree.
- Brillant solution: add all 0 gates to queue prior to BFS, and simultaneously spread from all of them level by level.

````python
class Solution: #OG, naive like number of islands
    def wallsAndGates(self, rooms: list[list[int]]) -> None:
        """
        Do not return anything, modify rooms in-place instead.
        """
        ROW, COL = len(rooms), len(rooms[0])
        directions = ((0,1),(1,0),(0,-1),(-1,0))
        queue = deque()
        INF = 2147483647
        
        for r in range(ROW):
            for c in range(COL):
                if rooms[r][c] == 0:
                    queue.append((r, c)) #append all gates
            
        while queue:
            n = len(queue)
            for _ in range(n):
                row, col = queue.popleft()
                for dr, dc in directions:
                    nrow, ncol = row+dr, col+dc
                    if 0<=nrow< ROW and 0<=ncol<COL and rooms[nrow][ncol] == INF:
                        rooms[nrow][ncol] = rooms[row][col]+1
                        queue.append((nrow, ncol))
        return rooms
````

#### [127. Word Ladder](https://leetcode.com/problems/word-ladder): 
Approach: BFS `unvisitedSet(wordList)`, get neighbor: replace each char w/ `ascii_lowercase`

````python
from string import ascii_lowercase
from collections import deque
class Solution:
    def ladderLength(self, beginWord: str, endWord: str, wordList: List[str]) -> int:
        # get the neighbors during processing each word. Word lenght is only 10 at max, replace 26 characters will be 260 operations at max for searching, 
        #use unvisited (init as set(wordList)) instead of visited -> can check if a newly generated neighbor is a valid, unused word
        queue = deque([beginWord])
        unvisited = set(wordList)
        wordLen = len(beginWord)
        level = 0
        while queue:
            n = len(queue)
            level+=1
            for _ in range(n):
                curWord = queue.popleft()
                #get neighbors for node
                for c in ascii_lowercase:
                    for i in range(wordLen):
                        if c != curWord[i]:
                            newWord = curWord[:i] + c + curWord[i+1:]
                            if newWord in unvisited: #valid neighbor, process
                                if newWord == endWord:
                                    return level+1 # level (parent) + 1 for newWord
                                unvisited.remove(newWord)
                                queue.append(newWord)
        return 0
````

#### [752. Open the Lock](https://leetcode.com/problems/open-the-lock)
- One state can transition to another state IF THEY DIFFER by one digit **(same as word ladder)**. 
- Use visitedSet is fine, create moving={"srcNum":[srcNum-1, srcNum+1]}, adjust for moving[0], moving[9]
        

````python
class Solution:    #Grady's cool code!
    #graph + min path: todo dijsktra
    def openLock(self, deadends: List[str], target: str) -> int:
        deadends = set(deadends)
        if target == "0000": return 0
        if target in deadends or "0000" in deadends: return -1

        q = deque(["0000"])
        visited = set(["0000"])
        level = 0
        moving = {str(i): [str(i-1), str(i+1)] for i in range(10)}
        moving["0"] = ["9", "1"]
        moving["9"] = ["8", "0"]
        while q:
            level+=1
            n = len(q)
            for _ in range(n): # process q's level
                cur = q.popleft() 
               
                for i in range(4):  # visit children of popCur
                    for replace in moving[cur[i]]: #replace is [prev, next]
                        newState = cur[:i] + replace + cur[i+1:]
                        if newState not in deadends and newState not in visited:
                            if newState == target: return level
                            visited.add(newState)
                            q.append(newState)
        
        return -1
                
````

#### [773. Sliding Puzzle](https://leetcode.com/problems/sliding-puzzle/description/)
- Flatten board to 1D to easily look for index of 0 arr.index(0). 1D form is: tuple/string - to be hashed in **visitSet**
- Same as the **Word Ladder**

````python
# Grady cool self solve, no assistance - 35 mins
class Solution:
    def slidingPuzzle(self, board: List[List[int]]) -> int:
        TARGET = (1,2,3,4,5,0)
        board = tuple([board[i][j] for i in range(len(board)) for j in range(len(board[0]))])

        if board == TARGET: return 0 #edge case. if board==target, return 0

        # BFS, transitition to neighbors with: direction[location]
        directions = [[1,3], [0,2,4], [1,5], [0,4], [1, 3, 5], [4, 2]]
        q = deque([board])
        visited = set([board]) # set of tuple states
        swapLevel = 0
               
        while q:
            n = len(q)
            swapLevel+=1
            for _ in range(n): #get all nodes at this level
                curState = q.popleft()  #[4,1,2,5,0,3] make tuple to list to modify

                # Get neighbors
                zeroIdx = curState.index(0)
                for dirIdx in directions[zeroIdx]:  
                    newState = list(curState) # Copy and swap indices
                    newState[zeroIdx], newState[dirIdx] = newState[dirIdx], newState[zeroIdx]
                    
                    newState = tuple(newState) #Check in visit, add to q&visit
                    if newState not in visited:
                        if newState == TARGET: return swapLevel
                        visited.add(newState)
                        q.append(newState)
        return -1
````

### Topological sort for Directed Graph
- Topological ordering: When orders of nodes matters/got directed (schedule, prerequisites, dependency....). One graph may have many orders.
- A Directed Acyclic Graph (DAG) has a cycle + have NO VALID topo order.
- Uses:
    - Find topological ordering in O(V+E).
##### <font color='yellow'>Toposort using DFS</font> 
- Init `topoIndex = n-1`, use PathSet for singlePath cycle checking, and visitSet (order index)
- Dfs `topoIndex = dfs(topoIndex, node)`:
    - topoIndex decreased by a series of node's unvisited neighbor. Because `dfs return topoIndex - 1`
    - After decrease, assign `topo[topoIndex] = node`
- Return topo

````python
#Util funcion turns adj to graph
from collections import defaultdict
def toGraph(n:int, adj:list, start:int)->dict:
    graph = defaultdict(list)
    #Create vertices
    for i in range(start,n+start):
        graph[i] = []
    #Adj
    for u,v in adj:
        graph[u].append(v)
      
    return graph
````

````python
def topsort(graph: dict) -> list:
    n = len(graph)
    visit = set()     # fully processed
    path = set()      # recursion stack
    topo = [-1] * n
    topoIndex = n - 1
     #In each dfs, go to neighbors -> assign to ordering, and DECREASE INDEX to assign to topo
    def dfs(topoIndex, node):        
        if node in path:      # 🔴 cycle, return -2 (idx == -1 is a valid, when all nodes processed)
            return -2       
        
        path.add(node)
        for child in graph[node]:
            if child not in visit:
                topoIndex = dfs(topoIndex, child) #1. MAIN POINT: Decrease assigning idx
                if topoIndex == -2: return -2
        
        path.remove(node)
        visit.add(node)       # 2. mark finished after child processing
        topo[topoIndex] = node # 3. Assign node to correct position in TOPOSORT
        return topoIndex - 1
    
    for node in graph:
        if node not in visit:
            topoIndex = dfs(topoIndex, node) #0. Decrease assigning idx to topo
            if topoIndex == -2: return []     # cycle detected
    return topo

graph = toGraph(6, [[1,0],[4,1],[4,5],[4,2],[0,2],[0,3],[2,3],[3,4],[3,5],[5,2]], start=0)
topsort(graph)
# graph2 = toGraph(6, [[1,5],[2,1],[3,1],[5,4],[0,4]], start=0) # loop for node in graph decrease topoIndex here. 
# topsort(graph2)
````

##### <font color='yellow'>Kahn's: BFS - in degree - Detect cycle</font> 

<img src="assets/kahn topo.png" width=500></img>   

Time: O(V+E) (each node enters Q once, each edge visited linearly).   
Space: O(V)

0. <font color="yellow">No visitSet. Topo = [], indegree={node:indegree}</font>
1. Init the queue with all nodes with indegree = 0
2. BFS via the <font color="yellow"> queue (only contains 0-degree nodes, ALL TIMES)</font>
- While q:
    - Only process one popNode, not all node in a level
    - Decrease indegree of all popnode.neighbors
        - If a neighbor's indegree == 0 => add to q
- If len(topo) != len(graph):
    - print("Detect cycle")

````python
from collections import deque
def topoBFS(graph: dict) -> list:
    q = deque() #
    topo = []
    indegree = [0]*(len(graph))

    #1. Get indegree of each vertex:
    for node in graph:
        for v in graph[node]:
            indegree[v] += 1
    
    #2. Add 0-indegree vertices to q:
    for i in range(0, len(indegree)):
        if indegree[i] == 0:
            q.append(i)
            
    #3. Similar to BFS, BUT process ONLY 1 popNode, not whole level (for_in range(len(q)))
    while q:
        popNode = q.popleft()
        topo.append(popNode) #Process a 0-indegree node, add to topo
        
        for child in graph[popNode]: #Visit child
            indegree[child] -= 1
            if indegree[child] == 0:
                q.append(child)
    if len(topo) < len(graph): # q finish earlier when there's cycle
        print("Detect cycle")
    return topo

#Test detect cycle
graph = toGraph(6, [[1,0],[4,1],[4,5],[4,2],[0,2],[0,3],[2,3],[3,5],[5,2]], start=0)
graph1 = toGraph(5,[[0, 1],[0, 2],[1, 2],[1, 3],[2, 3],[2,4]],start=0)
topoBFS(graph)
````

#### [444. Sequence Reconstruction](https://leetcode.com/problems/sequence-reconstruction)
Problem description:
- nums - permutation range n, `sequences` - subsequences of nums
- Return true if `nums` is the only shortest supersequence for `sequences`
```go
    nums = [1,2,3], sequences = [[1,2],[1,3],[2,3]] -> true
    nums = [1,2,3], sequences = [[1,2],[1,3]] -> false
```

````python
from collections import defaultdict
# Simple 2 loops, check every (num-1) is parent of (num)
# Determine unique supersequences: every i,need to be pointed to by i-1 (3 is child of 2, 2 is child of 1)
class Solution:
    def sequenceReconstruction(self, nums: list[int], sequences: list[list[int]]) -> bool:
        parentSet = defaultdict(set) #outdeg={3: {2,1}, 2: {1}, 4: {1,2,3}}
        for sequence in sequences:
            for i in range(len(sequence)-1):
                parentSet[sequence[i+1]].add(sequence[i]) 
        for i in range(len(nums)-1):
            if nums[i] not in parentSet[nums[i+1]]:
                return False
        return True
````

##### Kahn + Uniqueness:

- Topo NOT unique, when len(queue) > 1, at any time.
    - If there're more than one 0-indegree, we can pop any of them to obtain valid topo order -> **not unique**

````python
from collections import deque, defaultdict
class Solution:
    def sequenceReconstruction(self, nums: list[int], sequences: list[list[int]]) -> bool:
        # shortest supersequence: topo list of graph(sequences)
        #each sequence, 
        graph = defaultdict(set) #avoid add same neighbor from sequence list
        indegree = defaultdict(int)
        n = len(nums)
        q = deque()
        topo = []
        for sequence in sequences:
            for i in range(len(sequence)-1):
                if sequence[i+1] not in graph[sequence[i]]: #fast check tks to set()
                    graph[sequence[i]].add(sequence[i+1]) #graph={1: {2}, 2: {3, 4}, 3: {4}}
                    indegree[sequence[i+1]] += 1

        for i in range(1,n+1):
            if indegree[i] == 0:
                q.append(i)
                
        while q:
            if len(q) > 1: return False # 🟠UNIQUENESS
            popNode = q.popleft()
            topo.append(popNode)
            for child in graph[popNode]:
                indegree[child] -= 1
                if indegree[child] == 0:
                    q.append(child)
        return topo == nums #🟠CHECK IF THE UNIQUE PATH == NUMSb
sequences = [[1,2],[2,4],[2,3],[1,2,3,4]]
nums=[1,2,3,4]
Solution.sequenceReconstruction(Solution, nums, sequences)
````

**Description**:

2. Many courses can be taken at the same time.
- -> Return the minimum time complete all the courses.

**Intuition**:
- **Return longest time dependency path.**
- `pathTime(child) = time[child] + max(pathTime(parent))`

**Example**.  
<img src="./assets/Task Scheduling 2.png" height="350px"/>

````python
#DFS
class Solution:
    def minimumTime(self, n: int, relations: list[list[int]], time: list[int]) -> int:
        graph = defaultdict(list)
        pathTime = defaultdict(int) #{1: 3, 3: 8 (max(2,3)+5) } #time needed to finish 
        times = {i : time[i-1] for i in range(1,n+1)}
        
        for src, des in relations:
            graph[src].append(des)

        def dfsGetPathTime(node):
            if node in pathTime: # 1. Base case check if visited
                return pathTime[node]
            # 2. Get maxTime from children
            maxTime = max(dfsGetPathTime(child) for child in graph[node]) if graph[node] else 0
            pathTime[node] = maxTime + times[node] # 3. remember pathTime[node]
            return pathTime[node] 

        return max(dfsGetPathTime(node) for node in range(1, n+1)) # 0. Go thru all nodes
Solution.minimumTime(self=Solution, n=5, relations=[[1,5],[2,5],[3,5],[3,4],[4,5]], time=[1,2,3,4,5])
````

````python
class Solution: #Kahn BFS
    def minimumTime(self, n: int, relations: list[list[int]], time: list[int]) -> int:
        graph = defaultdict(list)
        indegree = defaultdict(int)
        q = deque()
        pathTime = defaultdict(int) #{1: 3, 3: 8 (max(2,3)+5) } #time needed to finish 
        times = {i : time[i-1] for i in range(1,n+1)}
        maxPath = max(time) #init as maxTime, there might be no relations and no q
        
        # 1. Init graph, indegree, pathTime <dict>, maxPath
        for src, des in relations:
            graph[src].append(des)
            indegree[des] += 1
            pathTime[src] = times[src]
            maxPath = max(pathTime[src], maxPath)
        # 2. 0-indegree to q
        for node in range(1, n+1):
            if indegree[node] == 0: q.append(node)
        #3. Kahn: Get one popNode, process children + indegree[child]
        while q:
            popNode = q.popleft()
            for child in graph[popNode]:
                pathTime[child] = max(pathTime[child], pathTime[popNode]+times[child])
                maxPath = max(maxPath, pathTime[child])

                indegree[child] -= 1
                if indegree[child] == 0:
                    q.append(child)
        
        return maxPath # no relations, but have some nodes in time[]
Solution.minimumTime(self=Solution, n=5, relations=[[1,5],[2,5],[3,5],[3,4],[4,5]], time=[1,2,3,4,5])
````

````python
# DFS
from collections import defaultdict
class Solution:
    def alienOrder(self, words: list[str]) -> str:
        nodeSet = set()
        for word in words:
            for c in word:
                nodeSet.add(c)
        topo = [""]* len(nodeSet)
        topoIndex = len(nodeSet) - 1
        visitSet = set()
        pathSet = set()
        graph = defaultdict(set)
        def getDependency(smallWord, bigWord) -> bool:  
            nonlocal graph                
            for j in range(min(len(smallWord), len(bigWord))): 
                if smallWord[j] != bigWord[j]: 
                    graph[smallWord[j]].add(bigWord[j])
                    return True
            # EDGE: 🔴 abc < ab -> INVALID  (No diff so far && len(sm) > len(big))
            if len(smallWord) > len(bigWord):
                return False    
            return True # same word

        for i in range(len(words) - 1): #get dependency
            if not getDependency(words[i], words[i+1]):
                return ""
        def dfsSinglePath(topoIndex, node) -> int:
            if node in pathSet: return -2 #Check cycle is most important
            if node in visitSet: return topoIndex

            pathSet.add(node)
            for child in graph[node]:
                topoIndex = dfsSinglePath(topoIndex, child)
                if topoIndex == -2: return -2
            
            pathSet.remove(node)
            visitSet.add(node) #mark visit AFTER process children, so children branch may visit node and detect cycle
            topo[topoIndex] = node
            return topoIndex - 1
        for node in nodeSet:
            if node not in visitSet: #Topo DFS, each visited once.
                topoIndex = dfsSinglePath(topoIndex, node) #make sure to maintain topoIdx properly
                if topoIndex == -2: return ""
        return "".join(topo)
````

### Shortest path algorithms

##### **1. BFS - SPFA** Simplest, can be used for negative edge
- Only need q(deque) and: `distance = {node: minDistToNode}`
* **Time:** O(V * E), in worst case, each time we update a new edge for visited node, we have to process all childNodes chain for that edge.
- Space: O(V+E)


````python
from collections import deque
from math import inf
#graph = {a:[(b,2),(c,3)], b: [(d:4)]}
def shortestPathSPFA(graph: list[list[tuple[int, int]]], root: int, target: int) -> int:
    q = deque([root])
    dist = {node : inf for node in range(len(graph))} #dist[node] = minDistanceToNode for nodes in graph
    dist[root] = 0
    while q:
        popNode = q.popleft() #every node, go to every child, update dist.
        for neighbor, weight in graph[popNode]:
            if dist[popNode] + weight < dist[neighbor]: # Need minUpdate in q and dist
                q.append(neighbor)
                dist[neighbor] = dist[popNode] + weight
    return dist[target] if dist[target] != inf else -1
````

- q = minHeap: [(1, **node1**), (2, node2), (3, **node1**)]
* **Use case:** Finding shortest paths in weighted graphs with non-negative edge weights.
* **Time complexity:** O(V log V + E) using a heap.

````python
from heapq import heappop, heappush
from math import inf
#Same as SPFA, but use heap instead of deque. Inspired by AlgoMonster Alien Dict: use heap for q to get min child
def shortestPathDijsktra(graph: list[list[tuple[int, int]]], root: int, target: int) -> int:
    q = [(0, root)] # curDistance, node
    dist = {node : inf for node in range(len(graph))} # node: minDistToNode
    dist[root] = 0
    while q:
        curDist, popNode = heappop(q)
         # 1. once node popped for the FIRST TIME, curDist guaranteed node's minDist S**
        if popNode == target: return curDist
        
        if curDist > dist[popNode]: continue # Act like lazy VISITED set => pass old worse version: e.g: pass heapNode(2,8), when dist[2]=3
        
        for child, weight in graph[popNode]: #2. Update children distance
            if dist[popNode] + weight < dist[child]: #Need min update in dist and q (maybe q again, to update child's children)
                dist[child] = dist[popNode] + weight
                heappush(q, (dist[child], child))
    
    return dist[target] if dist[target] != inf else -1 # always -1 ;)
# S**: when popNode - we done processed node, for forever. There can't be a smaller path reach node after this, because if there is a smaller path from root, it must have been popped before! -> Contradictory
````

#### [743. Network Delay Time](https://leetcode.com/problems/network-delay-time)
- Typical Dijsktra - exactly same as template, just return `Longest single path from k to a node.`
- Problems: Return the minimum time it takes for all the n nodes to receive the signal (IN PARALELL): **-> longest single path** 
    - If it is impossible, return -1.

````python
from math import inf
from heapq import heappush, heappop
class Solution:
    def networkDelayTime(self, times: List[List[int]], n: int, k: int) -> int:
        #minimum time for all nodes receive signals in parallel, is the longest path from k to anynode
        q = [(0, k)]
        dist = {i:inf for i in range(1,n+1)}
        dist[k] = 0
        graph = defaultdict(list)
        for src, des, w in times:
            graph[src].append((des,w))

        while q:
            curDist, popNode = heappop(q) 
            if curDist > dist[popNode]: continue
            for child, weight in graph[popNode]:
                if dist[popNode]+ weight < dist[child]:
                    dist[child] = dist[popNode]+ weight
                    heappush(q, (dist[child], child))
        
        longestPath = max(dist.values())
        return longestPath if longestPath != inf else -1
````

##### **4. Floyd-Warshall Algorithm**

* **Use case:** Finding shortest paths between all pairs of nodes in a weighted graph.
* **Time complexity:** O(V^3)


````python
def floyd_warshall(graph):
    """
    Finds the shortest paths between all pairs of nodes in a weighted graph
    using the Floyd-Warshall algorithm.

    Args:
        graph: A dictionary representing the graph where keys are nodes
               and values are dictionaries mapping neighbors to edge weights.

    Returns:
        A 2D list representing the shortest distances between all pairs of nodes.
    """
    nodes = list(graph.keys())
    num_nodes = len(nodes)
    distances = [[float('inf')] * num_nodes for _ in range(num_nodes)]

    # Initialize distances with edge weights and self-distances
    for i in range(num_nodes):
        for j in range(num_nodes):
            if i == j:
                distances[i][j] = 0
            elif nodes[j] in graph[nodes[i]]:
                distances[i][j] = graph[nodes[i]][nodes[j]]

    # Compute shortest paths
    for k in range(num_nodes):
        for i in range(num_nodes):
            for j in range(num_nodes):
                distances[i][j] = min(distances[i][j], distances[i][k] + distances[k][j])

    return distances
````

##### **5. A\* Search Algorithm**

* **Use case:** Finding shortest paths in weighted graphs, often used in pathfinding with a heuristic function to guide the search.
* **Time complexity:** Depends on the heuristic function and the graph structure.


````python
import heapq

def astar(graph, start, end, heuristic):
    """
    Finds the shortest path between two nodes in a weighted graph using
    the A* search algorithm with a heuristic function.

    Args:
        graph: A dictionary representing the graph where keys are nodes
               and values are dictionaries mapping neighbors to edge weights.
        start: The starting node.
        end: The ending node.
        heuristic: A function that estimates the distance from a node to the end node.

    Returns:
        A tuple containing the shortest distance and the path, or (float('inf'), None)
        if no path exists.
    """
    open_set = set([start])
    closed_set = set()
    g_score = {node: float('inf') for node in graph}
    g_score[start] = 0
    f_score = {node: float('inf') for node in graph}
    f_score[start] = heuristic(start, end)
    came_from = {}

    while open_set:
        current = min(open_set, key=lambda node: f_score[node])
        if current == end:
            # Reconstruct the path
            path = [current]
            while current in came_from:
                current = came_from[current]
                path.append(current)
            path.reverse()
            return g_score[end], path

        open_set.remove(current)
        closed_set.add(current)

        for neighbor, weight in graph[current].items():
            if neighbor in closed_set:
                continue
            tentative_g_score = g_score[current] + weight
            if neighbor not in open_set:
                open_set.add(neighbor)
            elif tentative_g_score >= g_score[neighbor]:
                continue

            came_from[neighbor] = current
            g_score[neighbor] = tentative_g_score
            f_score[neighbor] = g_score[neighbor] + heuristic(neighbor, end)

    return float('inf')
````

````python
class UnionFind: #Union by rank (review later)
    def __init__(self):
        self.id = {}
    def union(self, x, y): #rootx->rooty
        self.id[self.find(x)] = self.find(y)
    def find(self, node):
      parent = self.id.get(node, node)
      if parent != node: 
        self.id[node] = parent = self.find(parent) #path compress
      return parent
#Note: Kruskal needs edgeList [(src, des, weight)], nodeList; not graph: { src:[ (des, weight) ] }
def kruskal(n: int, edges: list[list[int, int, int]]) -> int: 
    #sort edge
    edges.sort(key= lambda x: x[2]) # element x in edge = [node1,node2,w]
    dsu = UnionFind()
    cost, edgeCount = 0, 0
    for x, y, weight in edges:
        if dsu.find(x) != dsu.find(y):
            dsu.union(x, y)
            cost += weight
            edgeCount += 1
            if edgeCount == n -1:
                break
    return cost if edgeCount == n-1 else -1
````

##### Prim ~ Dijkstra
- Time: O(V log V + E) using a heap on V
- Difference with Dijkstra:
    1. replace distance -> edgeWeight = [inf] * n
    2. In child loops, compare & assign: 
        - 👍`weight < edgeWeight[child]` 
    3. use visitedSet, to avoid add old nodes curCost to: total+=CurCost


````python
from heapq import heappop, heappush
from math import inf

def primMST(graph: list[list[tuple[int, int]]]) -> int:
    n = len(graph)
    
    q = [(0, 0)]  # (edgeCost, node) — start from node 0
    edgeWeight = {node: inf for node in range(n)}  # min edge cost to connect node
    visited = set()
    
    edgeWeight[0] = 0
    total = 0  # total MST cost

    while q:
        curCost, popNode = heappop(q)
        if popNode in visited:
            continue

        # Process popNode
        visited.add(popNode)
        total += curCost

        # Update neighbors
        for neighbor, weight in graph[popNode]:
            if neighbor not in visited and weight < edgeWeight[neighbor]:
                edgeWeight[neighbor] = weight
                heappush(q, (weight, neighbor))

    return total if len(visited) == n else -1
````

Optimized Prim: No heap. Time = O(N^2)
- For every node:
    - minCost = inf
    - Go to every other node to get minCost of unvisited node
    - mark minCost node, update total
    - Go to every node again, update egdeWeight for child not in visited 

````python
from math import inf
from typing import List

class Solution:
    def minCostConnectPoints(self, points: List[List[int]]) -> int:
        n = len(points)

        # edgeWeight[i] = minimum cost to connect point i to current MST
        edgeWeight = [inf] * n
        edgeWeight[0] = 0  # start from node 0

        visited = set()
        total = 0

        for _ in range(n):  # we must pick exactly n nodes
            # 1️⃣ Find unvisited node with smallest edgeWeight
            minCost = inf
            node = -1
            for i in range(n):
                if i not in visited and edgeWeight[i] < minCost:
                    minCost = edgeWeight[i]
                    node = i

            # 2️⃣ Add that node to MST
            visited.add(node)
            total += minCost

            # 3️⃣ Update edgeWeight for remaining unvisited nodes
            for j in range(n):
                if j not in visited:
                    dist = abs(points[node][0] - points[j][0]) + \
                           abs(points[node][1] - points[j][1])

                    if dist < edgeWeight[j]:
                        edgeWeight[j] = dist

        return total
````

##### Union by Rank - most optimal
Time: Find(): O(alpha(n)) (it will never practically speaking exceed O(4)); Union: O(1)
Very simple:
- Find: exact same
- Union: almost same, except:
    1. <font color = "yellow">rank</font> (child - 0 rank, root - highest rank)
    2. After find root, and root different:
        - <font color = "pink"> SMALLER-rank -> HIGHER-rank</font>
        - <font color = "orange"> SAME RANK: increment rank of the chosen parent (mergedRoot)</font> 

````python
class UnionFind: #Union by rank (review later)
    def __init__(self):
        self.id = {}
        self.rank = defaultdict(int)
    
    def union(self, x, y):
        xRoot, yRoot = self.find(x), self.find(y)
        if xRoot != yRoot: # 🔴 Core: merge to the higher rank root, if same rank, increment rank of the chosen parent
            if self.rank[xRoot] < self.rank[yRoot]: 
                self.id[xRoot] = yRoot
            else:                       # rank[xRoot] >= rank[yRoot]
                self.id[yRoot] = xRoot
                if self.rank[yRoot] == self.rank[xRoot]: 
                    self.rank[xRoot] += 1 # equal ranks -> increment chosen parent (xRoot)
    
    def find(self, node):
      parent = self.id.get(node, node)
      if parent != node: 
        self.id[node] = parent = self.find(parent)
      return parent
````

````python
class Solution:
    def deleteAndEarn(self, nums: List[int]) -> int:
        #Time: O(N+k - k is largestNum), space: O(N) - points
        # using DP maxPoints, where maxPoints[i] returns maxPoints we can obtain if we consider only upto i (this smart move only consider i and i-1, no need i+1 to the recurrence relation)
        points = defaultdict(int) #map num to totalPoints gain by num
        maxNum = 0 #max element in nums
        
        for num in nums:
            points[num] += num
            maxNum = max(maxNum, num)
        oneBack, twoBack = points[1], 0
        for i in range(2, maxNum+1):
            twoBack, oneBack = oneBack, max(oneBack, twoBack + points[i])
        return oneBack
````

````python
# Loop through left only, row by row:
def uniquePaths(m: int, n: int) -> int:
    grid = [1] * n #first row
    for _ in range(1, m):
        for j in range(1, n):
            grid[j] += grid[j-1] #inherits top, only add left
    return grid[-1]
# Loop through of top, left: grid[i][j] = grid[i][j-1] + grid[i-1][j]
#desc start from top-left, move down or right, to bottom-right corner
#same as climbing stairs thinking: if we can reach i from either 2 entities i-1 or i-2 => reach(i) = reach(i-1)+reach(i-2)
def uniquePaths(m: int, n: int) -> int:
    grid = [[0 for _ in range(n)] for _ in range(m)]
    for i in range(m):
        grid[i][0] = 1
    for i in range(n):
        grid[0][i] = 1
    for i in range(1, m):
        for j in range(1, n):
            grid[i][j] = grid[i][j-1] + grid[i-1][j] #i = left+top
    return grid[-1][-1]
        
````

````python
#memo + DFS
'''
text1[:i]   ... X
text2[:j]   ... X
'''
#lru_cache: be cautious about the param(keys) used
from functools import lru_cache
def longestCommonSubsequence(text1: str, text2: str) -> int:
    m, n = len(text1), len(text2)
    @lru_cache(maxsize=None)
    def lcs(i, j) -> int:           # LCS accumulated from FROM END to [i:j]
        if i == m or j == n: return 0
        
        if text1[i] == text2[j]: # Both prefixes end with the same char -> this must be added to the "upward" lcs
            return 1 + lcs(i+1, j+1)
        return max(lcs(i+1, j), lcs(i, j+1)) #get max lcs from either omit i or j
    return lcs(0, 0)
````

````python
# DFS + memo
from functools import lru_cache
'''
turns i -> j
after replace, will work at (i+1, j+1)
after delete i, will work at (i+1, j) # i deleted, so i+1 work with j
after insert j, will work at (i, j+1) # just insert a char before i, now i work with j+1

Why use memo
From (i, j) you can reach (i+1, j+1) in two different ways:
Path A: insert then delete | Path B: delete then insert
(i, j) -> (i, j+1) -> (i+1, j+1) | (i, j) -> (i+1, j) -> (i+1, j+1)
Same (i+1, j+1), different paths ⇒ dfs(i+1, j+1) multiple times.

'''

def minDistance(word1: str, word2: str) -> int:
    m,n = len(word1), len(word2)
    @lru_cache(maxsize=None)
    def dfs(i, j):
        if i == m and j == n:
            return 0
        if i == m: #if one string reaches end, we either 1 insertion or 1 deletion
            return n - j
        if j == n:
            return m-i
        if word1[i] == word2[j]:
            return dfs(i+1, j+1)
        return 1 + min(dfs(i+1, j+1), dfs(i, j+1), dfs(i+1, j))
    return dfs(0,0)
word1 = "intentionaa"
word2 = "exction"
minDistance(word1, word2)
````

````python
#DFS memoi
from functools import lru_cache

def delete_string(costs: list[int], s1: str, s2: str) -> int:
    def getCost(char): return costs[ord(char) - ord('a')]
        
    n, m = len(s1), len(s2)
    from functools import lru_cache

    @lru_cache(None)
    def dfs(i: int, j: int) -> int:
        # Base cases: delete remaining characters
        if i == m:
            return sum(getCost(char) for char in s2[j:])
        if j == n:
            return sum(getCost(char) for char in s1[i:])

        
        if s1[i] == s2[j]: # Characters match - move next one point for both with 0 cost
            return dfs(i + 1, j + 1)

        # Delete from s1 or s2, pick cheaper option, and move forward to next char.
        delete_s1 = getCost(s1[i]) + dfs(i + 1, j)
        delete_s2 = getCost(s1[j]) + dfs(i, j + 1)
        return min(delete_s1, delete_s2)

    return dfs(0,0)
````

````python
#2D rolling up from 3D
import math
#T: O(n × op1 × op2), S: O(op1 × op2)
class Solution:
    def minArraySum(self, nums, d, op1, op2):
        INF = float('inf')

        # dp[j][k] = min sum using j op1 and k op2 so far
        dp = [[INF]*(op2+1) for _ in range(op1+1)]
        dp[0][0] = 0

        for num in nums:
            tempDP = [[INF]*(op2+1) for _ in range(op1+1)]
            # Key different from 3D, where each loop, we find dp[i][j][k] from past dp[i-1][j-1][k-1]. 
            # In 2D rolling up, each i,k loop, we push to future dp[i][j+1][k] from current dp[j][k], and conditional check instead of j,k >0, it's j,k<len
            for j in range(op1+1):
                for k in range(op2+1):
                    base = dp[j][k]
                    if base == INF: continue
                    # 1) no operation
                    tempDP[j][k] = min(tempDP[j][k], base + num)
                    # 2) op1
                    if j < op1:
                        tempDP[j+1][k] = min(tempDP[j+1][k], base + (num + 1)//2)
                    # 3) op2
                    if k < op2 and num >= d:
                        tempDP[j][k+1] = min(tempDP[j][k+1], base + (num - d))
                    # 4a) op1 then op2
                    if j < op1 and k < op2:
                        v1 = (num + 1)//2
                        if v1 >= d:
                            tempDP[j+1][k+1] = min( tempDP[j+1][k+1], base + (v1 - d))
                    # 4b) op2 then op1
                    if j < op1 and k < op2 and num >= d:
                        v2 = num - d
                        tempDP[j+1][k+1] = min(tempDP[j+1][k+1], base + (v2 + 1)//2)
            dp = tempDP

        return min(min(row) for row in dp)
````

````python
# Count: count[i] > 1 means extra -> push to i+1, cascade forward
def minIncrementForUnique(nums: list[int]) -> int:
    # create a counter arr manually -> auto sorted by value, as index
    # maxVal can incremented up to maxVal + n => len(count) = maxVal+n
    count = [0] * (max(nums)+len(nums))
    moves = 0
    for num in nums:
        count[num] += 1
    for i in range(len(count)):
        if count[i] > 1: # last count[-1] WON'T > 1
            moves += count[i] - 1
            count[i+1] += count[i] - 1
            count[i] = 1
    return moves
````

````python
# 3. DSU O(n · log(n))
# min increments unique → Union-Find next-available slot - root of tree.
# find(x) returns free slot, parent[slot]=slot+1 -> set parent = point forward
def minIncrementForUnique(nums):
    parents = dict()
    def find(x): #find "what's the next free slot at or after x"
        parentX = parents.get(x, x) #return x if x has no parent
        if parentX != x:
            parents[x] = parentX = find(parentX) #path compression
        return parentX
    moves = 0
    for num in nums:
        nextSlot = find(num)
        moves += nextSlot - num
        parents[nextSlot] = nextSlot+1 # point used slot to nexSlot+1
    return moves
````

#### [134. Gas Station](https://leetcode.com/problems/gas-station/)
Given: Running circular with tankGas = gas[i] = cost[i]
Identify: ith starting point:
Running tank prefix → if tank<0 → immediately reset tank = 0, start=i+1

```go

tank[start -> i] just becomes NEG => Reset tank & newstart=i+1 (*)

                                       'final'
[5      -7          3        1   -4       2        4]
[5      -2(!)   1(reset)     2   -2   0(reset)   6]

(*): All points between [start -> i] CANT be startCandidate because that would make tank more NEG(-)
```

````python
class Solution:
    def canCompleteCircuit(self, gas: List[int], cost: List[int]) -> int:
        if sum(gas) < sum(cost): return -1
        tank = 0 # accumulative gas
        start = 0
        for i in range(len(gas)):
            tank += gas[i] - cost[i]
            if tank < 0:  # 🔥 tank[start -> i] just becomes NEG => reset tank immedately!
                tank = 0 
                start = i+1 # 🔥 new candidateStart = i + 1
        return start
````

# Divide & Conquer
Main idea: 
1. Split main problem -> 2 smaller subs, where already recursively solved
2. Merge 2 subs to solve main problems.

Note: can be used to optimize problems

````python
def mergeSort(arr: list[int]) -> list[int]:
    n = len(arr)
    # 0. Base
    if n < 2:  
        return arr
        
    # 1. Divide, and resolve recursively
    mid = len(arr)//2
    leftSub = mergeSort(arr[:mid])
    rightSub = mergeSort(arr[mid:])
    
    # 2. Merge 2 sorted arrays (classic, use 2 pointers)
    l, r, res = 0, 0, []
    while l < len(leftSub) and r < len(rightSub):
        if  leftSub[l] < rightSub[r]:
            res.append(leftSub[l])
            l+=1
        else:
            res.append(rightSub[r])
            r+=1
    res.extend(leftSub[l:]) 
    res.extend(rightSub[r:])
    return res
````

#### [218. The Skyline Problem](http://leetcode.com/problems/the-skyline-problem)
Given: retangles buildings [x, y, height], return skyline points (points when height changes)

**APPROACH for all solutions**: 
- At each unique point, get max height (rightEdgeHeight = 0). 
- Add res, when maxHeight changes.

**Divide and Conquer approach**: better read solution
1. Divide buildings to 2 halfs; leftSkyline and rightSkyline

    - Base case len==1, return 2-point skyline:
        - `return [[left, height], [right, 0]] `
2. Merge skylines:
    - If height changed, Add to res the blend of: [soonerPosition, maxHeight]

        <img src="./assets/218. The Skyline Problem1.jpeg" height="500px"/>

````python
# Read solution is the best way to understand.
class Solution:
    def getSkyline(self, buildings: list[list[int]]) -> list[list[int]]:
        # 0. Base case: single building -> 2 skyline points (left, height) and (right, 0)
        if not buildings: return []
        if len(buildings) == 1:
            l, r, h = buildings[0]
            return [[l, h], [r, 0]]
        
        # 1. Divide
        mid = len(buildings) // 2
        left  = self.getSkyline(buildings[:mid])
        right = self.getSkyline(buildings[mid:])
        
        # 2. Merge two skylines (like merge sorted arrays, but track max height from each side)
        return self.merge(left, right)
    def merge(self, leftHalf, rightHalf):
        leftCurHeight, rightCurHeight = 0, 0 # Current maxHeight of leftHalf/rightHalf, it helps preserve proper maxHeight
        l, r, res = 0, 0, []
        while l < len(leftHalf) and r < len(rightHalf):
            # 1. Compare sooner position, update curMaxHeight of chosen side
            if leftHalf[l][0] < rightHalf[r][0]: 
                pos, leftCurHeight = leftHalf[l]
                l+=1
            elif leftHalf[l][0] > rightHalf[r][0]:
                pos, rightCurHeight = rightHalf[r]
                r+=1 
            else: # same edge -> process both curMaxHeight and pointers l, r
                pos, leftCurHeight = leftHalf[l]
                _, rightCurHeight = rightHalf[r]
                l+=1
                r+=1
            # 2. Add to res if curHeigh changed:
            maxHeight = max(leftCurHeight, rightCurHeight)
            if not res or res[-1][1] != maxHeight:
                res.append([pos, maxHeight])
        res.extend(leftHalf[l:])
        res.extend(rightHalf[r:])
        return res
                
            
````

**Optimized BF: Heap to get maxHeight quick at each point i**

**Approach**:
```go
    I. Get maxHeight quick at point i with 'activeHeap=[(buildingHeight, buildingRightEnd)]': 

        1. Add building containing point i at heap
        2. Lazy delete EXPIRED topHeaps (bc only topHeaps affect maxHeight[point]) (EXP: rightEnds <= point)
        3. maxHeight[i] always = topHeap

    II. Same as Brute-Force
```
<img src="./assets/218. The Skyline Problem2.jpg" height="300px"/>

````python
from heapq import heappush, heappop
class Solution:
    def getSkyline(self, buildings: list[list[int]]) -> list[list[int]]:
        pointSet = sorted(list(set(point for building in buildings for point in building[:2])))
        res = []
        j = 0 # building index -> each building visit once
        activeHeap = [] #maxHeap
        # I. get maxHeights at point i quick with activeHeap=[(height, rightEnd)]
        for point in pointSet: 
            # 1. Add all available building at point to heap
            while j < len(buildings) and buildings[j][0] == point:
                _, rightEnd, height = buildings[j]
                heappush(activeHeap, (-height, rightEnd))
                j+=1
                
            # 2. Remove topHeap expired to achive proper maxHeight[point] with topHeap (if expired but not topHeap => don't care bc it's not maxHeight[point] anyway)
            while activeHeap and activeHeap[0][1] <= point:
                heappop(activeHeap)
            curMaxHeight = -activeHeap[0][0] if activeHeap else 0
            # 3. Only append curMaxHeight to res if diff from prevMaxheight in res)
            if not res or res[-1][1] != curMaxHeight:
                res.append((point, curMaxHeight))
        return res
Solution().getSkyline(buildings)
````

#### [315. Count of Smaller Numbers after Self | Number of Swaps to Sort | Algorithm Swap](https://algo.monster/problems/count_of_smaller_numbers_after_self)

- Number of Swaps to Sort = sum([Count of Smaller Numbers after Self](#sorting))
- 2 other approaches: Binary Indexed Tree & Segment Tree: help do range query in logN

**Approach: Merge sort**
```go
1. Divide arr to 2 half 'LEFTSUB' & `RIGHTSUB`, each sorted recursively
2. While SortMerging (2 sorted arr - 2 pointers l, r):
    - ⭐️ Can get 'countSmallers of LEFTSUB' nums FROM `RIGHTSUB`, by:
        - When curLeft being merged:
            -> countSmallers[curLeft] += r (r: curLeft''s smallerRights processed)
```

``` go
Merging count[curLeft] += r_index visual:

[1,2,5,6,9] + [3,4,7,8] 

l                r
1  2  5  6  9 |  3  4  7  8   (count=[0,0,0,0,0,0,0,0,0])
count[1] += r=0
count[2] += r=0

      l          r
1  2  5  6  9 |  3  4  7  8
                  r→1, r→2

      l  l             r
1  2  5  6  9 |  3  4  7  8   (count=[0,0,2,2,0,0,0,0,0])
count[5] += r=2
count[6] += r=2

            l          r  r
1  2  5  6  9 |  3  4  7  8         (count=[0,0,2,2,4,0,0,0,0])
                        r→3, r→4
count[9] += r=4

```

````python
class Solution: 
    def countSmaller(self, nums: list[int]) -> list[int]:
        count = [0] * len(nums) #count of rightSmallers
        # nums = [(idx, val)]
        def mergeSort(nums: list[list[int]]) -> list[int]:
            if len(nums) <= 1: 
                return nums
            mid = len(nums) // 2
            leftSub = mergeSort(nums[:mid])
            rightSub = mergeSort(nums[mid:])
            return merge(leftSub, rightSub)
            
        def merge(leftSub: list[list[int]], rightSub: list[list[int]]) -> list[list[int]]:
                nonlocal count
                l, r, res = 0, 0, []
                # add to count[] when curRight < curLeft
                while l < len(leftSub) or r < len(rightSub):
                    if l >= len(leftSub )or (r < len(rightSub) and rightSub[r][1] < leftSub[l][1]): # if curRight < curLeft 
                        res.append(rightSub[r])
                        r+=1 # ⭐️ R_index == curRightSmallersCount of cur L_index (bc these prev rights < cur next L_index)
                    else:  # When curLeft merging: 
                        count[leftSub[l][0]] += r # count[curLeft] += all previous smallerRights count
                        res.append(leftSub[l])
                        l+=1
                return res

        mergeSort(list(enumerate(nums))) # [(idx, val) in nums]
        return count
````

# Sorting

**[Best Visualizer](https://visualgo.net/en/sorting)**
``` go
- Bubble sort: "pairs" story
    - We have n(n-1) pairs. Each pair EITHER CORRECT or INVERSION (i<j, but num[i]>num[j])
    - 1 swap fix only 1 INVERSION of the specific two swapped num (use visualizer)
        - When one largerLeft SWAPPING, it FIXING ONLY its relationship with smallerRights, NOT AFFECT OTHER RELATIONSHIPS
> 'Total swaps in bubble sorts = Sum([counts of smallers-on-the-right of eachNum])'
```
E.g: arr = [5, 2, 6, 1, 3, 7, 8, 4]

- smallers[5] = 4; smallers[2] = 1; smallers[6] = 3;  smallers[1] = 0,...
- -> num[5] needs to do 4 swaps to fix 4 inversions no matter what, and it doesnt affect anyone


````python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):  # sort the biggest at the end (n-i)th first - by just swapping adjacents
        for j in range(n - i - 1):  # only scan unsorted part [0, n-i-1)
            if arr[j] > arr[j + 1]:  # inversion found → swap adjacent, remove exactly 1 inversion
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

def insertion_sort(arr):
    n = len(arr)
    for i in range(1, n):  # assume [0,i-1] sorted, insert arr[i] into correct position
        key = arr[i]        # save current element to insert
        j = i - 1
        while j >= 0 and arr[j] > key:  # RIGHTSHIFT anything > key to make room
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key    # insert key into correct sorted position

def selection_sort(arr):
    n = len(arr)
    for i in range(n):      # find the ith min, put to ith
        min_index = i
        for j in range(i + 1, n):  # scan unsorted part to find minimum
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]  # swap min to front of unsorted
````

````python
import random
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def __init__(self, head: Optional[ListNode]):
        self.head = head
    def getRandom(self) -> int:
        cur = self.head
        chosen = -1
        i = 1
        while cur:
            if random.random() < 1/i: # randomProb < (1/i)Prob
                chosen = cur.val
            cur = cur.next
            i +=1
        return chosen
````

Skip adding these points 'v' '^' in mat1

````python
def multiply(mat1: list[list[int]], mat2: list[list[int]]) -> list[list[int]]:
    rowMat1, colMat1Share, colMat2 = len(mat1), len(mat1[0]), len(mat2[0]) #colMat1 == RowMat2 by rules
    res = [[0]* colMat2 for _ in range(rowMat1)] 
    # normally loop: row->col->shared(each cell)
    # But to optimized '0' cells: row -> shared (skip '0' cells) -> col
    for i in range(rowMat1):
        for k in range(colMat1Share):
            if rowMat1[i][k] == 0: continue
            for j in range(colMat2):
                res[i][j] += mat1[i][k]*mat2[k][j]
    return res
````

````python
def image_blur(image, radius):
    rows, cols = len(image), len(image[0])
    result = [[0] * cols for _ in range(rows)]
    
    for i in range(rows):
        for j in range(cols):
            neighbors = []
            
            # Collect valid neighbors within the specified radius, excluding (i, j) itself
            for k in range(max(0, i - radius), min(rows, i + radius + 1)):
                for l in range(max(0, j - radius), min(cols, j + radius + 1)):
                    if k == i and l == j:
                        continue  # Skip the center cell
                    neighbors.append(image[k][l])
            
            # Calculate mean of neighbors and apply the formula
            if neighbors:
                mean_neighbors = sum(neighbors) // len(neighbors)
                result[i][j] = (image[i][j] + mean_neighbors) // 2
            else:
                # No neighbors, so the pixel remains the same
                result[i][j] = image[i][j]
                
    return result
image = [[9, 6], [3,0]]

image_blur(image, 1)
````

#### OA2

Given an infinite integer number line, you would like to build some blocks and obstacles on it. Specifically, you have to implement code which supports two types of operations:

[1, x] - builds an obstacle at coordinate x along the number line. It is guaranteed that coordinate x does not contain any obstacles when the operation is performed.

[2, x, size] - checks whether it's possible to build a block of size size which ends immediately before x on the number line. For example, if x = 6 and size = 2, this operation checks coordinates 4 and 5. Produces "1" if it is possible, i.e. if there are no obstacles at the specified coordinates, or produces "0" otherwise. Please note that this operation does not actually build the block, it only checks whether a block can be built.

Given an array of operations containing both types of operations described above, your task is to return a binary string representing the outputs for all [2, x, size] operations.

Example

For

operations = [[1, 2],

              [1, 5],

              [2, 5, 2],

              [2, 6, 3],

              [2, 2, 1],

              [2, 3, 2]]

the output should be solution(operations) = "1010".

Explanation:

Let's consider all operations:

[1, 2] - builds an obstacle at coordinate 2.

[1, 5] - builds an obstacle at coordinate 5.

[2, 5, 2] - checks and produces "1" as it is possible to build a block occupying coordinates 3 and 4.

[2, 6, 3] - checks and produces "0" as it is not possible to build a block occupying coordinates 3, 4, and 5, because there is an obstacle at coordinate 5.

[2, 2, 1] - checks and produces "1" as it is possible to build a block occupying coordinate 1.

[2, 3, 2] - checks and produces "0" as it is not possible to build a block occupying coordinates 1 and 2 because there is an obstacle at coordinate 2.

So the output is "1010". 


````python
from sortedcontainers import SortedList

def solution(operations):
    obstacles = SortedList()
    result = []

    for op in operations:
        if op[0] == 1:
            # Operation [1, x]: Place an obstacle at x
            obstacles.add(op[1])
        elif op[0] == 2:
            # Operation [2, x, size]: Check if we can place a block of 'size' ending at 'x-1'
            x, size = op[1], op[2]
            start = x - size
            # Find the first obstacle that is >= start
            idx = obstacles.bisect_left(start)
            
            # If there's an obstacle before 'x' in the range [start, x-1], it's not possible
            if idx < len(obstacles) and obstacles[idx] < x:
                result.append('0')
            else:
                result.append('1')

    return ''.join(result)

# Example usage
operations = [[1, 2], [1, 5], [2, 5, 2], [2, 6, 3], [2, 2, 1], [2, 3, 2]]
print(solution(operations))  # Output: "1010"
````

````python
def count_numbers_with_total(X, Y):
    # Convert X to a string for easier manipulation
    X_str = str(X)

    # Memoization for digit DP
    def digit_dp(pos, temp, total):
        # Base case: all positions processed
        if pos == len(X_str):
            return 1 if total == 0 else 0
        
        # If total becomes negative, it's invalid
        if total < 0:
            return 0
        
        # Memoize states (position, temp, total)
        if (pos, temp, total) in memo:
            return memo[(pos, temp, total)]
        
        # Determine the limit for the current digit
        limit = int(X_str[pos]) if temp else 9
        result = 0
        
        # Try all digits from 0 to `limit`
        for digit in range(0, limit + 1):
            result += digit_dp(
                pos + 1,
                temp and (digit == limit),
                total - digit
            )
        
        # Store in memo
        memo[(pos, temp, total)] = result
        return result

    # Initialize memoization dictionary
    memo = {}
    
    # Start digit DP from position 0, temp bound, and the target digit sum
    result = digit_dp(0, True, Y)
    return result if result > 0 else -1


# Example usage
inputNum1 = 100  # X
inputNum2 = 10   # Y
print(count_numbers_with_total(inputNum1, inputNum2)) 
````

````python
# 2571. Minimum Operations to Reduce an Integer to 0
def minOperations(n: int) -> int:
    #Greedy process bin rep of n bit by bit from right to left
    ops = 0 #counts of total operations needed
    count1 = 0 #count of total consecutive 1-bits seen

    while n > 0: 
        if n & 1: # if current bit == 1
            count1 += 1 #update consecutive count1
        elif count1: # cur bit == 0, had some count1 > 0
        #if cnt == 1, take 1 eliminate op; reset cnt = 0
        #if cnt > 1, take 1 add op to "collapse" to "higher single 1-bit", cnt = 1 for new carried bit
            ops += 1
            count1 = 0 if count1 == 1 else 1
        n >>= 1 #Right shift to check next bit
    # Handle all remain consecutive 1bits after loop, because the loop keeps accumulating many count1, it only collapsed when see 0
    if count1 == 1:
        ops += 1 # 1 elimination op
    elif count1 > 1:
        ops += 2 # collapse to higher single 1 bit, then eliminate
    return ops
````

42. Trapping Rain Water 2 POINTERS
```python
class Solution:
    def trap(self, height):
        """
        :type height: List[int]
        :rtype: int
        """
        # Step 1: Initialize Pointers and Variables
        left, right = 0, len(height) - 1  # Two pointers at the start and end of the array
        ans = 0  # Stores total trapped water
        left_max, right_max = 0, 0  # Track max height seen so far from left and right
        
        # Step 2: Two-Pointer Traversal
        while left < right:
            # Step 2a: Process the Shorter Side First
            if height[left] < height[right]:
                # Step 2b: Update left_max and Calculate Water Trapped
                left_max = max(left_max, height[left])  # Update max height seen on the left
                ans += left_max - height[left]  # Water trapped at index left
                left += 1  # Move left pointer to the right
            else:
                # Step 2c: Update right_max and Calculate Water Trapped
                right_max = max(right_max, height[right])  # Update max height seen on the right
                ans += right_max - height[right]  # Water trapped at index right
                right -= 1  # Move right pointer to the left
        
        return ans  # Return the total trapped water
```

1. **Two-Pointer Approach**:
   - Instead of using a stack, this solution **uses two pointers (`left` and `right`)** to track the boundaries.
   - It efficiently calculates trapped water in **O(n) time** without extra space.

2. **Core Idea**:
   - **Water trapped at any index depends on the shorter boundary** (either `left_max` or `right_max`).
   - The algorithm **processes the shorter side first** because water is limited by the shorter boundary.

3. **How It Works**:
   - **If `height[left] < height[right]`**: Process the left side, because water is trapped **only up to `left_max`**.
   - **Otherwise**: Process the right side, because water is trapped **only up to `right_max`**.
   - **Update `left_max` and `right_max`** as we move the pointers.

4. **Efficiency**:
   - **Time Complexity:** `O(n)` (Each element is processed once)
   - **Space Complexity:** `O(1)` (Only a few variables are used, no extra memory)


````python
class Solution:
    def vowelStrings(self, words: List[str], queries: List[List[int]]) -> List[int]:
        # Step 1: Initialize Variables
        n = len(words)  # Number of words
        Prefix = [0] * (n + 1)  # Prefix array to store cumulative counts of valid words
        vowels = {'a', 'e', 'i', 'o', 'u'}  # Set of vowels for quick lookup

        # Step 2: Build the Prefix Array
        for i in range(n):
            # Carry forward the previous count
            Prefix[i + 1] = Prefix[i]
            # Check if the word starts and ends with a vowel
            if words[i][0] in vowels and words[i][-1] in vowels:
                Prefix[i + 1] += 1  # Increment the count if valid

        # Step 3: Process Queries Using the Prefix Array
        res = []  # Result array to store answers for each query
        for query in queries:
            # Calculate the count of valid words in the range [query[0], query[1]]
            res.append(Prefix[query[1] + 1] - Prefix[query[0]])

        # Step 4: Return the Results
        return res
````

````python
#2524. Maximum Frequency Score of a Subarray - slide window
from collections import Counter

class Solution:
    def max_frequency_score(self, nums, k):
        # Define the modulus for the calculations as per the problem statement, which is 10^9 + 7
        mod = 10**9 + 7

        # Create a Counter for the first 'k' elements in the list
        counter = Counter(nums[:k])

        # Calculate the initial score by raising 'k' to the power of the frequency count, modulo 'mod'
        current_score = sum(pow(num, frequency, mod) for num, frequency in counter.items()) % mod

        # Set the 'answer' to the initial score, it will keep track of the highest score
        answer = current_score

        # Start iterating from the 'k'th element in 'nums'
        i = k
        while i < len(nums):
            # 'previous_num' is the number which will be excluded from the current window
            # 'new_num' is the number which will be included in the current window
            previous_num, new_num = nums[i - k], nums[i]

            # Only process if we're examining a new number; otherwise, skip re-calculation
            if previous_num != new_num:
                # Increase the score by the power of the count of the new number, if it already exists
                if counter[new_num]:
                    current_score += (new_num - 1) * pow(new_num, counter[new_num], mod)
                else:
                    current_score += new_num

                # Decrease the score by the power of the count of the previous number, if necessary
                if counter[previous_num] > 1:
                    current_score -= (previous_num - 1) * pow(previous_num, counter[previous_num] - 1, mod)
                else:
                    current_score -= previous_num

                # Ensure the current score does not exceed 'mod'
                current_score %= mod

                # Update the counter for the new window
                counter[new_num] += 1
                counter[previous_num] -= 1

                # Update the answer if the current score is higher
                answer = max(answer, current_score)

            # Move to the next element
            i += 1

        # Return the highest score found
        return answer
````

```python
class Solution:
    def findMedianSortedArrays(self, nums1: List[int], nums2: List[int]) -> float:
        # Step 1: Ensure nums1 is the smaller array
        n1, n2 = len(nums1), len(nums2)
        # Swap arrays if nums1 is larger to ensure efficiency in binary search
        if n1 > n2:
            return self.findMedianSortedArrays(nums2, nums1)

        # Step 2: Initialize variables for binary search
        low, high = 0, n1  # Binary search range in nums1
        left = (n1 + n2 + 1) // 2  # The total elements on the left side of the split
        total = n1 + n2  # Total number of elements in nums1 and nums2

        # Step 3: Perform Binary Search
        while low <= high:
            # Calculate partitions for nums1 and nums2
            mid1 = (low + high) // 2  # Partition index in nums1
            mid2 = left - mid1  # Partition index in nums2

            # Assign boundaries for nums1 and nums2
            l1 = nums1[mid1 - 1] if mid1 > 0 else float('-inf')  # Left of partition in nums1
            l2 = nums2[mid2 - 1] if mid2 > 0 else float('-inf')  # Left of partition in nums2
            r1 = nums1[mid1] if mid1 < n1 else float('inf')  # Right of partition in nums1
            r2 = nums2[mid2] if mid2 < n2 else float('inf')  # Right of partition in nums2

            # Step 3a: Check if the partitions are correct
            if l1 <= r2 and l2 <= r1:
                # If total length is odd, the median is the max of left partitions
                if total % 2 == 1:
                    return max(l1, l2)
                # If total length is even, the median is the average of max left and min right
                else:
                    return (max(l1, l2) + min(r1, r2)) / 2

            # Step 3b: Adjust partitions if not correct
            elif l1 > r2:
                # Move left in nums1
                high = mid1 - 1
            else:
                # Move right in nums1
                low = mid1 + 1

        # Step 4: Return 0 (should never reach here)
        return 0  # This line is unreachable
```

---

 **Summary of Key Concepts**

1. **Binary Search on the Smaller Array**:
   - To minimize time complexity, we always perform binary search on the smaller array (`nums1`).
   - If `nums1` is larger, we swap the inputs.

2. **Core Idea**:
   - **Divide both arrays into two halves** such that:
     - All elements in the left half are ≤ all elements in the right half.
   - Use binary search to find the correct partition index.

3. **Partitioning and Median**:
   - If the total number of elements is odd, the median is the maximum element of the left half.
   - If even, the median is the average of the maximum of the left half and the minimum of the right half.

4. **Efficiency**:
   - **Time Complexity:** `O(log(min(n1, n2)))` because we perform binary search on the smaller array.
   - **Space Complexity:** `O(1)` as no extra space is used.
